```python
import matplotlib.pyplot as plt
import numpy as np


x=np.array([5,12])
y=np.array([6,14])
plt.subplot(2,3,1)
plt.title("sales")
plt.xlabel("x label",color='b')
plt.ylabel("y label",color='r')
plt.plot(x,y,marker='o')


x=np.array([2,5,2,5])
y=np.array([5,2,5,2])
plt.subplot(2,3,2)
plt.title("sales")
plt.plot(x, color='y',linestyle='dashed',marker='*')
plt.plot(y, color='b',marker="*",linewidth='2.5',mfc='y')


y = np.array([3, 8, 1, 10])
plt.subplot(2,3,3)
plt.plot(y, marker = 'o', mec = 'k',mfc='y')

plt.subplots_adjust(hspace=0.5) 

x=np.array([0,5,10,15])
plt.subplot(2,3,4)
plt.title('sales')
plt.plot(x)
plt.grid()


x=np.array([3,6,8,4,19,46,76,34,23])
y=np.array([7,3,8,4,5,23,43,17,26])
colors=np.array(["red","green","blue","yellow","pink","black","orange","purple","beige"])
sizes = np.array([20,50,100,200,60,90,10,80,75])
plt.subplot(2,3,5)
plt.scatter(x,y,c=colors,s=sizes)


x=np.array(['A','B','C'])
y=np.array([5,3,7])
plt.subplot(2,3,6)
plt.bar(x,y,color='r',width=0.1,)
plt.show()

```

```python
import matplotlib.pyplot as plt
import numpy as np



y = np.array([35, 20, 20, 25])
mylabels = ["Apples", "Bananas", "Cherries", "Dates"]
myexplode = [0.1, 0, 0., 0]
mycolors=["red","green","blue","yellow"]

plt.pie(y,labels=mylabels,startangle=90,explode=myexplode,shadow=True,colors=mycolors)
plt.legend()
```
