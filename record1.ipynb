```python
# e using function return true if 2 list have atleast one common value
def common_member(list1, list2):
    for item in list1:
        if item in list2:
            return True
    return False

list1 = [4,7,8,2]
list2 = [3,4,2,5]

print(common_member(list1, list2))
```
